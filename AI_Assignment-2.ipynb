```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
digits = load_digits()
fig,axes = plt.subplots(5,2,figsize = (18,5))
for i ,ax in enumerate(axes .flat):
    ax.imshow (digits.images[i],cmap='grey')
    ax.set_title(f"Digit: {digits.target[i]}")
    ax.axis('off')

            
plt.tight_layout()
plt.show()
```

AI_Assignment2_Q1 : Write a python program to import iris dataset from scikitlearn  and convert it into pandas dataframe and display the shape of the dataset.

```python

from sklearn.datasets import load_iris
import pandas as pd


iris = load_iris()


df = pd.DataFrame(iris.data, columns=iris.feature_names)


df["target"] = iris.target


print("Iris Dataset:")
print(df)


print("\nShape of the dataset:", df.shape)
```

AI_Assignment2_Q3 : WAP to count the no. of samples available in iris datasets and digit datasets .

```python

from sklearn.datasets import load_iris, load_digits


iris = load_iris()
digits = load_digits()


iris_samples = len(iris.data)
digits_samples = len(digits.data)


print("Number of samples in Iris dataset:", iris_samples)
print("Number of samples in Digits dataset:", digits_samples)
```
