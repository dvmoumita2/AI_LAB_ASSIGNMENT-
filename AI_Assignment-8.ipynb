```python
import math

def minimax(depth, node_index, is_maximizing, values, max_depth):

    
    if depth == max_depth:
        return values[node_index]

    if is_maximizing:
        return max(
            minimax(depth + 1, node_index * 2, False, values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, False, values, max_depth)
        )

    else:
        return min(
            minimax(depth + 1, node_index * 2, True, values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, True, values, max_depth)
        )



values = [3, 5, 2, 9, 12, 5, 23, 23]

max_depth = 3

result = minimax(0, 0, True, values, max_depth)

print("The optimal value is:", result)
```

```python
import math

def minimax(state, depth, maximizing_player):

    if game_over(state):
        return evaluate(state)

    if maximizing_player:
        best = float('-inf')

        for move in get_moves(state):
            new_state = make_move(state, move, 'O')
            score = minimax(new_state, depth + 1, False)
            best = max(best, score)

        return best

    else:
        best = float('inf')

        for move in get_moves(state):
            new_state = make_move(state, move, 'X')
            score = minimax(new_state, depth + 1, True)
            best = min(best, score)

        return best


def game_over(state):
    
    for row in state:
        if row[0] == row[1] == row[2] and row[0] != ' ':
            return True

    
    for i in range(3):
        if state[0][i] == state[1][i] == state[2][i] and state[0][i] != ' ':
            return True

    
    if state[0][0] == state[1][1] == state[2][2] and state[0][0] != ' ':
        return True

    if state[0][2] == state[1][1] == state[2][0] and state[0][2] != ' ':
        return True

    
    for row in state:
        if ' ' in row:
            return False

    return True


def evaluate(state):
    
    for row in state:
        if row[0] == row[1] == row[2]:
            if row[0] == 'O':
                return 10
            elif row[0] == 'X':
                return -10

    
    for i in range(3):
        if state[0][i] == state[1][i] == state[2][i]:
            if state[0][i] == 'O':
                return 10
            elif state[0][i] == 'X':
                return -10

    if state[0][0] == state[1][1] == state[2][2]:
        if state[0][0] == 'O':
            return 10
        elif state[0][0] == 'X':
            return -10

    if state[0][2] == state[1][1] == state[2][0]:
        if state[0][2] == 'O':
            return 10
        elif state[0][2] == 'X':
            return -10

    return 0


def get_moves(state):
    moves = []

    for i in range(3):
        for j in range(3):
            if state[i][j] == ' ':
                moves.append((i, j))

    return moves


def make_move(state, move, player):
    new_state = [row[:] for row in state]

    i, j = move
    new_state[i][j] = player

    return new_state




state = [
    ['X', 'O', 'X'],
    [' ', 'O', ' '],
    [' ', ' ', ' ']
]

print("Current Game State:")

for row in state:
    print(row)

best_value = minimax(state, 0, True)

print("\nBest value for Computer (O):", best_value)
```

```python
import math


def alpha_beta(depth, node_index, maximizing_player, values, alpha, beta, max_depth):

    
    if depth == max_depth:
        return values[node_index]

    if maximizing_player:
        best = float('-inf')

        for i in range(2):
            value = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                False,
                values,
                alpha,
                beta,
                max_depth
            )

            best = max(best, value)
            alpha = max(alpha, best)

            
            if beta <= alpha:
                break

        return best

    else:
        best = float('inf')

        for i in range(2):
            value = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                True,
                values,
                alpha,
                beta,
                max_depth
            )

            best = min(best, value)
            beta = min(beta, best)

            
            if beta <= alpha:
                break

        return best



values = [3, 5, 2, 9, 12, 5, 23, 23]

max_depth = 3

alpha = float('-inf')
beta = float('inf')


optimal_score = alpha_beta(
    0, 0, True, values, alpha, beta, max_depth
)

print("Leaf Node Values:", values)
print("Optimal Score for MAX player:", optimal_score)
```
