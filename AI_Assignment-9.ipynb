Problem 1: Develop a Python program to implement a Rule-Based Expert System for Student Performance Evaluation .The system should take inpits such as attendance, marks, assignment performance,and internal asseccment and use IF THEN rules to classify the student as Excellent , Good, or Poor.  

```python
print("STUDENT PERFORMANCE EVALUATION SYSTEM")
attendance = float(input("Enter attendance performance: "))
marks = float(input("Enter marks percentage: "))
assignment = float(input("Enter assignment performance percentage: "))
internal = float(input("Enter internal assessment percentage: "))
if (attendance >= 90 and marks >= 85 and assignment >= 85 and internal >= 85):
    performance = "Excellent"
elif (attendance >= 75 and marks >= 60 and assignment >= 60 and internal >= 60):
    performance = "Good"
else:
    performance = "Poor"
print("Student Performance:", performance)
```

Problem 2: Develop a Python program to implement a Rule-Based Expert System for MEDICAL DIAGNOSIS. The system should except symptoms such as fever, cough, headache and sore throat  and use IF THEN rules to identify a possible disease. 

```python
print("MEDICAL DIAGNOSIS EXPERT SYSTEM")
fever = input("Do you have fever? (yes/no): ")
cough = input("Do you have cough? (yes/no): ")
headache = input("Do you have headache? (yes/no): ")
sore_throat = input("Do you have sore throat? (yes/no): ")
if (fever == "yes" and cough == "yes" and headache == "yes" and sore_throat == "yes"):
    disease = "Flu"
elif (fever == "yes" and cough == "yes" and sore_throat == "yes"):
    disease = "Common Cold"
elif (fever == "yes" and headache == "yes"):
    disease = "Fever-related illness"
elif (cough == "yes" and sore_throat == "yes"):
    disease = "Throat Infection"
else:
    disease = "No specific disease identified"
print("Possible Disease:", disease)
```
